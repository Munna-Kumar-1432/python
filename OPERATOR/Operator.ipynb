# 4. Logical Operators
- Used to combine conditional statements.

| Operator | Meaning                                        |
| -------- | ---------------------------------------------- |
| `and`    | Returns True if both conditions are True       |
| `or`     | Returns True if at least one condition is True |
| `not`    | Reverses the result                            |


| A     | B     | A and B |
| ----- | ----- | ------- |
| True  | True  | True    |
| True  | False | False   |
| False | True  | False   |
| False | False | False   |


```python
age = 20
has_license = True

print(age >= 18 and has_license)
```

| A     | B     | A or B |
| ----- | ----- | ------ |
| True  | True  | True   |
| True  | False | True   |
| False | True  | True   |
| False | False | False  |


```python
marks = 85
sports_quota = False

print(marks >= 90 or sports_quota)
```

```python
True and True   # True
True and False  # False

True or False   # True
False or False  # False

not True        # False
not False       # True
```

| Expression | Result |     |
| ---------- | ------ | --- |
| `5 & 3`    | `1`    |     |
| `5         | 3`     | `7` |
| `5 ^ 3`    | `6`    |     |
| `~5`       | `-6`   |     |
| `~-5`      | `4`    |     |
| `5 << 1`   | `10`   |     |
| `5 << 2`   | `20`   |     |
| `20 >> 2`  | `5`    |     |
| `-5 << 1`  | `-10`  |     |
| `-5 >> 1`  | `-3`   |     |


```python
print(5 & 1)   # 1
print(5 | 1)   # 5
print(5 ^ 1)   # 4
print(~5)      # -6
print(5 << 2)  # 20
print(20 >> 2) # 5
```
